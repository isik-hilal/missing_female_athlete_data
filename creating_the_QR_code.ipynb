# Creating the QR code of this Repo

I decided it is a nice idea to add this QR code generation here as a useful thing :)

```python
# please uv pip install "qrcode[pil]" pillow before

import qrcode

url = "https://github.com/isik-hilal/missing_female_athlete_data"

qr = qrcode.QRCode(
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=12,
    border=4,
)

qr.add_data(url)
qr.make(fit=True)

img = qr.make_image(
    fill_color="#6A0DAD",  # Purple
    back_color="#FFFFFF",  # White
)

img.save("missing_female_data_qr.png")

print("QR code saved as missing_female_data_qr.png")
```
